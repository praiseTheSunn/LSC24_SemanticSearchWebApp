```python
import pandas as pd
```

```python
df = pd.read_csv("LHE_keyframe_to_time_mapping.csv")
df.head()
```

```python
df['id'] = df.index + 1
df['image_id'] = df.apply(lambda row: f"LHE/{row['video_id']}/{row['keyframe_id']}", axis=1)
df['start_time'] = df['timeframe']
df['end_time'] = df['timeframe']
df = df[['id', 'image_id', 'video_id', 'keyframe_id', 'start_time', 'end_time']]
df.head()
```

```python
df.to_csv("./metadata_v1.csv", index=False)
```
