### Add OCR

```python
import pandas as pd
```

```python
df = pd.read_csv("./metadata_v1.csv")
df.head()
```

```python
# for each row, map the image_id to the link of a json path containing ocr results and set the ocr text to column 'ocr'
import os
import json

def get_ocr_text(image_id):
    json_path = os.path.join("/home/hlmquan/LSC24_SemanticSearchWebApp/backend_update/data/vbs25_v3c/ocr", image_id.replace("V3C/", "") + ".json")
    if os.path.exists(json_path):
        with open(json_path, "r") as f:
            data = json.load(f)
            rec_texts = data["rec_texts"]
            return " ".join(rec_texts)
    else:
        return ""
    

df["ocr"] = df["image_id"].apply(get_ocr_text)
df.head()
```
