```python
import pandas as pd
```

```python
df1 = pd.read_csv("./MVK_keyframe_to_time_mapping.csv")
df2 = pd.read_csv("./MVK2_keyframe_to_time_mapping.csv")

# concatenate the two dataframes
df = pd.concat([df1, df2], ignore_index=True)
df.head()
```

```python
# add id column starting from 1
# add image_id column: "MVK/{video_id}/{keyframe_id:05d}.jpg"
df['id'] = df.index + 1
df['image_id'] = df.apply(lambda row: f"MVK/{row['video_id']}/{row['keyframe_id']:05d}", axis=1)
df['start_time'] = df['timeframe']
df['end_time'] = df['timeframe']
df = df[['id', 'image_id', 'video_id', 'keyframe_id', 'start_time', 'end_time']]

df.head()
```

```python
df.to_csv("./metadata_v1.csv", index=False)
```
