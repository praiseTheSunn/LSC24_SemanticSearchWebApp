```python
df = pd.read_csv('metadata_v6.csv')
```

```python
df.head()
```

```python
def image_id_to_context_id(image_id):
    image_id = image_id.split('/')[-1]
    # return image_id[:11] + '_' + ('0' if image_id[11] < '3' else '1')
    return image_id[:8] + '_' + str(int(image_id[9:11]) // 2)

df['context_id'] = df['image_id'].apply(image_id_to_context_id)

# random print
df.sample(5)
```

```python
df.to_csv('metadata_v6.csv', index=False)
```
