```python
# read this file
# /home/hlmquan/LSC24_SemanticSearchWebApp/backend_update/database/color/color_output.db
import sqlite3

# to csv
conn = sqlite3.connect('/home/hlmquan/LSC24_SemanticSearchWebApp/backend_update/database/color/color_output.db')

# find table name
# table name 'color_detection'
table_name = 'color_detection'
df_color = pd.read_sql_query(f'SELECT * FROM {table_name}', conn)
df_color.to_csv('color_detection.csv', index=False)
df_color['image_id'] = df_color['image_id'].apply(lambda x: f"{x[:6]}/{x[6:8]}/{x}")
df_color.head()
```

```python
# add to metadata_v6.csv (df)
df_merged = pd.merge(df, df_color, on=['image_id'], how='left')
df_merged.head()
```
