```python
import numpy as np 
import settings
import importlib
importlib.reload(settings)

embeddings = np.load(settings.stfm_embedding_path)
offset = 0
batch_size = 512
length = len(keyframe_paths)

while offset < length:
    print(offset)    
    embds = embeddings[offset : min(offset + batch_size, length)]
    urls = keyframe_paths[offset : min(offset + batch_size, length)]
    urls = ["http://34.124.236.208/img_lsc/" + url[21:-4] + '.webp' for url in urls]
    urls = [url.replace('\\', '/') for url in urls]
    data = [{"url": url, "embedding": embedding} for url, embedding in zip(urls, embds)]
    res = client.insert(collection_name="clip_embeddings_cosine", data=data)
    print("Done inserting embeddings for image: ", offset + min(offset + batch_size, length) - 1)
    offset += batch_size
```

```python
import numpy as np

embedding_paths = sorted(glob.glob(os.path.join(settings.beit3_embedding_path, "*.npy")))
offset = 0

for path in embedding_paths:           # each day
    # embeddings cho 1 ngay 
    embds = np.load(path)
    print("Len of embds: ", len(embds))
    # url cho cac anh tuong ung trong 1 ngay, duoc chuyen thanh dang webp   
    urls = keyframe_paths[offset:offset + len(embds)]
    urls = ["http://34.124.236.208/img_lsc/" + url[21:-4] + '.webp' for url in urls]
    urls = [url.replace('\\', '/') for url in urls]
    # tao data de insert vao collection
    data = [{"url": url, "embedding": embedding} for url, embedding in zip(urls, embds)]
    # insert
    res = client.insert(collection_name="beit3_embeddings", data=data)    
    print("Done inserting embeddings for day: ", path)
    # di chuyen offset trong keyframe_paths
    offset = offset + len(embds)
```
