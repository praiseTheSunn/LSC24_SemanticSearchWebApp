```python
import torch
import torch.nn.functional as F 

PREFIXES = ["L05"]

for ebd_file in sorted(os.listdir(CLIP_EMBEDDINGS_DIR)):
    prefix = ebd_file.split("_")[0]
    if prefix not in PREFIXES:
        continue
# for ebd_file in MISSING_FILES:
    if ebd_file.endswith('.npy'):
        ebd_path = f"{CLIP_EMBEDDINGS_DIR}/{ebd_file}"
        video_id = ebd_file[:-4]
        embeddings = np.load(ebd_path)
        image_links = metadata_df[metadata_df['video_id'] == video_id].index.to_list()
        # print(image_links)
        # print(embeddings.shape[0], len(image_links))
        if embeddings.shape[0] == len(image_links):            # if ebd is found but no metadata found then the data is invalid -> no upload
            data = []
            for i in range(len(image_links)):
                image_link = image_links[i]
                embedding = embeddings[i]
                # print(f"Before Shape: {embedding.shape}")
                embedding_tensor = torch.tensor(embedding)
                embedding_normalized = F.normalize(embedding_tensor, dim=-1).numpy().astype('float32')
                # print(f"After Shape: {embedding_normalized.shape}")
                data.append({
                    "url": image_link,
                    "embedding": embedding_normalized
                })
        else:
            data = []
            count = 0
            print()
            print(f"Invalid data: {video_id}")
            for i in range(len(image_links)):
                image_link = image_links[i]
                available = metadata_df.loc[image_link, 'image_available']
                if available == 1:
                    data.append({
                        "url": image_link,
                        "embedding": embeddings[count].astype('float32')
                    })
                    count += 1               
                else:
                    print(image_link)

            
        print()
        res = client.insert(collection_name="vbs25_clips", data=data)
        print(res)            
        print(video_id)
        print(len(data), len(embeddings))
```
